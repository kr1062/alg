```python
import bisect

def max_profit(x, r, min_dist=5):
    points = sorted(zip(x, r))
    xs = [p[0] for p in points]
    rs = [p[1] for p in points]
    n = len(xs)

    dp = [0] * (n + 1)
    take = [False] * (n + 1)

    for i in range(1, n + 1):
        idx = i - 1
        target = xs[idx] - min_dist - 1e-9
        j = bisect.bisect_right(xs, target)

        profit_take = rs[idx] + dp[j]
        profit_skip = dp[i - 1]

        if profit_take >= profit_skip:
            dp[i] = profit_take
            take[i] = True
        else:
            dp[i] = profit_skip

    chosen_coords = []
    i = n
    while i > 0:
        if take[i]:
            chosen_coords.append(xs[i - 1])
            idx = i - 1
            target = xs[idx] - min_dist - 1e-9
            j = bisect.bisect_right(xs, target)
            i = j
        else:
            i -= 1

    chosen_coords.reverse()
    return dp[n], chosen_coords

x = [6, 7, 12, 14]
r = [5, 6, 5, 1]
profit, coords = max_profit(x, r)
print(profit)
print(coords)
```

Шаги жадного алгоритма:

1.   Выбираем узел с максимальным весом → v2 (вес 5)
2.   Добавляем v2 в S
3.   Удаляем v2 и его соседей (v1 и v3)
4.   Граф становится пустым
5.   Результат: S = {v2}, сумма = 5

Оптимальное решение:
Лучшее независимое множество: {v1, v3}, сумма = 4 + 4 = 8
Проблема в том, что выбрав узел с максимальным весом (v2), мы удаляем его соседей (v1 и v3), которые вместе дают бóльшую сумму, чем один v2.

Жадный алгоритм "слеп" к тому, что два меньших узла могут быть лучше одного большого, если они не конфликтуют между собой.
