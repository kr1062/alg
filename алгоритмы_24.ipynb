```python
from collections import deque

def build_adj_from_edges(n, edges):
    adj = [[] for _ in range(n)]
    for u, v, w in edges:
        adj[u].append((v, w))
        adj[v].append((u, w))
    return adj

def max_on_path_in_tree(n, tree_edges, src, dst):
    adj = build_adj_from_edges(n, tree_edges)
    parent = [-1] * n
    parent_edge = [None] * n
    stack = [src]
    parent[src] = src
    while stack:
        u = stack.pop()
        if u == dst:
            break
        for v, w in adj[u]:
            if parent[v] == -1:
                parent[v] = u
                parent_edge[v] = (u, v, w)
                stack.append(v)
    if parent[dst] == -1:
        return None
    cur = dst
    max_w = -1
    while cur != src:
        u, v, w = parent_edge[cur]
        if w > max_w:
            max_w = w
        cur = parent[cur]
    return max_w

def is_tree_still_mst(n, tree_edges, new_edge):
    v, w, c = new_edge
    max_w = max_on_path_in_tree(n, tree_edges, v, w)
    if max_w is None:
        return False, "Дерево несвязно"
    if c < max_w:
        return False, f"Не является MST: вес нового ребра {c} < максимум на пути {max_w}"
    else:
        return True, f"Остаётся MST: вес нового ребра {c} >= максимум на пути {max_w}"

n = 4
T = [(0,1,1), (1,2,1), (2,3,10)]

new_e1 = (0,3,6)
ok1, msg1 = is_tree_still_mst(n, T, new_e1)
print(msg1)

new_e2 = (0,3,12)
ok2, msg2 = is_tree_still_mst(n, T, new_e2)
print(msg2)

new_e3 = (0,3,10)
ok3, msg3 = is_tree_still_mst(n, T, new_e3)
print(msg3)
```

```python
#b
def build_adj_from_edges(n, edges):
    adj = [[] for _ in range(n)]
    for u, v, w in edges:
        adj[u].append((v, w))
        adj[v].append((u, w))
    return adj

def find_path_and_max(n, tree_edges, src, dst):
    adj = build_adj_from_edges(n, tree_edges)
    parent = [-1]*n
    parent_edge = [None]*n
    stack = [src]
    parent[src] = src
    while stack:
        u = stack.pop()
        if u == dst:
            break
        for v, w in adj[u]:
            if parent[v] == -1:
                parent[v] = u
                parent_edge[v] = (u, v, w)
                stack.append(v)
    if parent[dst] == -1:
        return None
    path_edges = []
    cur = dst
    while cur != src:
        e = parent_edge[cur]
        path_edges.append(e)
        cur = parent[cur]
    path_edges.reverse()
    max_w = -1
    max_idx = -1
    for i, (_, _, w) in enumerate(path_edges):
        if w > max_w:
            max_w = w
            max_idx = i
    return path_edges, max_w, max_idx

def update_mst_if_needed(n, tree_edges, new_edge):
    v, w, c = new_edge
    res = find_path_and_max(n, tree_edges, v, w)
    if res is None:
        return tree_edges, False, "Дерево несвязно"
    path_edges, max_w, max_idx = res
    if c < max_w:
        edge_to_remove = path_edges[max_idx]
        new_tree = []
        removed = False
        ru, rv, rw = edge_to_remove
        for (a, b, wt) in tree_edges:
            if not removed and ((a == ru and b == rv and wt == rw) or (a == rv and b == ru and wt == rw)):
                removed = True
                continue
            new_tree.append((a, b, wt))
        new_tree.append((v, w, c))
        return new_tree, True, f"Заменено ребро {edge_to_remove} (вес {max_w}) на новое ребро {(v,w,c)}"
    else:
        return tree_edges, False, "Дерево остаётся MST (новое ребро не лучше максимума на пути)"

n = 4
T = [(0,1,1), (1,2,1), (2,3,10)]

new_e1 = (0,3,6)
new_T1, changed1, msg1 = update_mst_if_needed(n, T, new_e1)
print(msg1)
print(new_T1)

new_e2 = (0,3,12)
new_T2, changed2, msg2 = update_mst_if_needed(n, T, new_e2)
print(msg2)
print(new_T2)

new_e3 = (0,3,10)
new_T3, changed3, msg3 = update_mst_if_needed(n, T, new_e3)
print(msg3)
print(new_T3)
```
