```python
def min_base_stations(houses, radius=4):
    if not houses:
        return 0
    houses = sorted(houses)
    stations = 0
    i = 0
    n = len(houses)

    while i < n:
        station_pos = houses[i] + radius
        stations += 1

        while i < n and houses[i] <= station_pos + radius:
            i += 1
    return stations


def get_stations_positions(houses, radius=4):
    houses = sorted(houses)
    stations = []
    i = 0
    n = len(houses)

    while i < n:
        pos = houses[i] + radius
        stations.append(pos)
        while i < n and houses[i] <= pos + radius:
            i += 1

    return stations

houses = [1, 3, 6, 10, 15]
print(f"Дома: {houses}")
print(f"Минимальное количество вышек: {min_base_stations(houses, 4)}")
print(f"Позиции вышек: {get_stations_positions(houses, 4)}")
```
