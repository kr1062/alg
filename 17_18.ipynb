17_1) Чтобы потраченная сумма была минимальной, нужно покупать лицензии в порядке убыванмя коэффициентов их роста

```python
def optimal_purchase_order(rates):
    licenses = [(i, rate) for i, rate in enumerate(rates)]
    licenses.sort(key=lambda x: x[1], reverse=True)
    return [idx for idx, _ in licenses]

def calculate_total_cost(rates, order):
    total = 0
    for t, idx in enumerate(order):
        total += 100 * (rates[idx] ** (t-1))
    return total

rates = [1.5, 2.0, 1.2]

print("Скорости роста:", rates)

order = optimal_purchase_order(rates)
print("Оптимальный порядок покупки (индексы):", order)
print("Оптимальный порядок покупки (скорости):", [rates[i] for i in order])

worst_cost = calculate_total_cost(rates, [0,1,2])
print(f"Стоимость при исходном порядке: {worst_cost}")
cost = calculate_total_cost(rates, order)
print(f"Минимальная общая стоимость:{cost}")
```

17_2) Если цена оборудования снижается каждый месяц, то в первую очередь нужно продавать то оборудование, у которого rj меньше, так как такие виды теряют свою стоимость быстрее

18

```python
def is_subsequence(S, S_prime):
    i = 0
    for event in S:
        if i < len(S_prime) and event == S_prime[i]:
            i += 1
        if i == len(S_prime):
            return True
    return i == len(S_prime)

S = ["Amazon", "Yahoo", "eBay", "Yahoo", 'Google', "Oracle"]
S_prime = ["Yahoo", "eBay", "Yahoo", "Oracle"]

print('Является подпоследовательностью' if is_subsequence(S, S_prime) else 'Не является подпоследовательностью')
```
