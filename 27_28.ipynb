```python
#27
def find_peak(arr):
    left, right = 0, len(arr) - 1

    while left < right:
        mid = (left + right) // 2

        if arr[mid] < arr[mid + 1]:
            left = mid + 1
        else:
            right = mid

    return left


A = [1, 3, 5, 7, 9, 12, 15, 14, 10, 8, 6, 4, 2]
peak_index = find_peak(A)
print(f"Peak index: {peak_index}")
print(f"Peak value: {A[peak_index]}")
```

```python
#28
def best_days(prices):
    if len(prices) < 2:
        return None

    buy_day = 0
    sell_day = 1
    min_price_day = 0
    max_profit = prices[1] - prices[0]

    for i in range(1, len(prices)):
        if prices[i] - prices[min_price_day] > max_profit:
            max_profit = prices[i] - prices[min_price_day]
            buy_day = min_price_day
            sell_day = i

        if prices[i] < prices[min_price_day]:
            min_price_day = i

    if max_profit <= 0:
        return None

    return buy_day + 1, sell_day + 1


p = [9, 1, 5]
result = best_days(p)
if result:
    print(f"Покупать в день {result[0]}, продавать в день {result[1]}")
else:
    print("Прибыль невозможна")
```
